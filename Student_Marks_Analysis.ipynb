```python

import matplotlib.pyplot as plt

students = {}

n = int(input("Enter number of students: "))

for i in range(n):
    name = input("Enter student name: ")
    marks = int(input("Enter marks out of 100: "))
    students[name] = marks

print("\n--- Student Marks Report ---")

for name, marks in students.items():
    if marks >= 90:
        grade = "A"
    elif marks >= 75:
        grade = "B"
    elif marks >= 50:
        grade = "C"
    else:
        grade = "Fail"

    print(name, marks, grade)

print("Total:", sum(students.values()))
print("Average:", round(sum(students.values()) / len(students), 2))
print("Highest:", max(students.values()))

plt.bar(students.keys(), students.values())
plt.title("Student Marks Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.ylim(0, 100)
plt.show()
```
